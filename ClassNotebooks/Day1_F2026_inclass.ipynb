While not as common, since we are in a science class, we will want to use a lot of scientific constants. Conveniently, there is a specific package for this (https://docs.scipy.org/doc/scipy/reference/constants.html):

```python
import scipy.constants as sc
```

Note the above "import [    ] as [    ]" commands in all of these cells. This is done to make your coding easier. For example, you could just do "import numpy" and then every time you wanted to use it in your code you would have to type out "numpy". By importing as a shorter name, you can make the code shorter and easier to work with. The numpy and pyplot names above are very standard and it will be easier to follow online examples if you do that. scipy.constants is less standard, so I just chose to use "sc"

Go to the web page for the scipy.constants documentation. It contains all constants. You call a constant as the following:

```python
print(sc.speed_of_light)
```
